```python
import json
from pathlib import Path

# File paths
INPUT_FILE = Path("../data/final_quiz_data.json")
OUTPUT_FILE = Path("../data/final_quiz_data_translated.json")

print("✓ Setup complete")
```

## Load Quiz Data

```python
# Load quiz data
with open(INPUT_FILE, 'r', encoding='utf-8') as f:
    quiz_data = json.load(f)

print(f"Loaded {len(quiz_data)} tests")
```

## Restructure Data - Add Arabic and French Answer Fields

```python
# Counters
signs_updated = 0
priorities_updated = 0
questions_updated = 0

# Process each test
for test_name, test_data in quiz_data.items():
    sections = test_data.get('sections', {})
    
    # Update Road Signs
    for sign in sections.get('road_signs', []):
        # Move correct_answer to answer_arabic
        if 'correct_answer' in sign:
            sign['answer_arabic'] = sign.pop('correct_answer')
            sign['answer_french'] = "###"
            signs_updated += 1
        # If already has answer_arabic, just add french placeholder if missing
        elif 'answer_arabic' in sign and 'answer_french' not in sign:
            sign['answer_french'] = "###"
            signs_updated += 1
    
    # Update Priorities
    for priority in sections.get('priorities', []):
        # Move correct_answer to answer_arabic
        if 'correct_answer' in priority:
            priority['answer_arabic'] = priority.pop('correct_answer')
            priority['answer_french'] = "###"
            priorities_updated += 1
        # If already has answer_arabic, just add french placeholder if missing
        elif 'answer_arabic' in priority and 'answer_french' not in priority:
            priority['answer_french'] = "###"
            priorities_updated += 1
    
    # Update General Questions
    for question in sections.get('general_questions', []):
        # Move correct_answer to answer_arabic
        if 'correct_answer' in question:
            question['answer_arabic'] = question.pop('correct_answer')
            question['answer_french'] = "###"
            questions_updated += 1
        # If already has answer_arabic, just add french placeholder if missing
        elif 'answer_arabic' in question and 'answer_french' not in question:
            question['answer_french'] = "###"
            questions_updated += 1

print(f"Updated structure:")
print(f"  Road Signs: {signs_updated}")
print(f"  Priorities: {priorities_updated}")
print(f"  General Questions: {questions_updated}")
print(f"  TOTAL: {signs_updated + priorities_updated + questions_updated}")
```

## Save Restructured Data

```python
# Save to new file
with open(OUTPUT_FILE, 'w', encoding='utf-8') as f:
    json.dump(quiz_data, f, ensure_ascii=False, indent=2)

print(f"✓ Saved restructured data to: {OUTPUT_FILE}")
print(f"\nFile size: {OUTPUT_FILE.stat().st_size / 1024:.1f} KB")
```

## Verification - Check Structure

```python
# Reload and verify
with open(OUTPUT_FILE, 'r', encoding='utf-8') as f:
    verified_data = json.load(f)

# Count items with new structure
signs_with_structure = 0
priorities_with_structure = 0
questions_with_structure = 0

for test_name, test_data in verified_data.items():
    sections = test_data.get('sections', {})
    
    for sign in sections.get('road_signs', []):
        if 'answer_arabic' in sign and 'answer_french' in sign:
            signs_with_structure += 1
    
    for priority in sections.get('priorities', []):
        if 'answer_arabic' in priority and 'answer_french' in priority:
            priorities_with_structure += 1
    
    for question in sections.get('general_questions', []):
        if 'answer_arabic' in question and 'answer_french' in question:
            questions_with_structure += 1

print(f"Verification Results:")
print(f"  Road Signs with answer_arabic & answer_french: {signs_with_structure}")
print(f"  Priorities with answer_arabic & answer_french: {priorities_with_structure}")
print(f"  Questions with answer_arabic & answer_french: {questions_with_structure}")
print(f"  TOTAL: {signs_with_structure + priorities_with_structure + questions_with_structure}")
```
